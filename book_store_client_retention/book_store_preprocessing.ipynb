```python
import numpy as np
from sklearn import preprocessing

raw_data = np.loadtxt('Audiobooks_data.csv', delimiter=',')
unscaled_inputs = raw_data[:, 1:-1]
targets = raw_data[:, -1]
```

### Balancing

```python
num_one_targets = int(np.sum(targets))
zero_targets_cnt = 0
ind_rmv = []

for i in range(targets.shape[0]):
    if targets[i] == 0:
        zero_targets_cnt += 1
        if zero_targets_cnt > num_one_targets:
            ind_rmv.append(i)

unscaled_inputs_equal_priors = np.delete(unscaled_inputs, ind_rmv, axis=0)
targets_equal_priors = np.delete(targets, ind_rmv, axis=0)
```

### Standardizing, shuffling, splitting and saving in .npz file

```python
scaled_inputs = preprocessing.scale(unscaled_inputs_equal_priors)
```

```python
shuffled_ind = np.arange(scaled_inputs.shape[0])
np.random.shuffle(shuffled_ind)

shuffled_inputs = scaled_inputs[shuffled_ind]
shuffled_targets = targets_equal_priors[shuffled_ind]
```

```python
samples_cnt = shuffled_inputs.shape[0]

train_samples_cnt = int(0.8 * samples_cnt)
validation_samples_cnt = int(0.1 * samples_cnt)
test_samples_cnt = samples_cnt - train_samples_cnt - validation_samples_cnt

train_inputs = shuffled_inputs[:train_samples_cnt]
train_targets = shuffled_targets[:train_samples_cnt]

validation_inputs = shuffled_inputs[train_samples_cnt:train_samples_cnt + validation_samples_cnt]
validation_targets = shuffled_targets[train_samples_cnt:train_samples_cnt + validation_samples_cnt]

test_inputs = shuffled_inputs[train_samples_cnt + validation_samples_cnt:]
test_targets = shuffled_targets[train_samples_cnt + validation_samples_cnt:]
```

```python
np.savez('Audiobook_data_train', inputs=train_inputs, targets=train_targets)
np.savez('Audiobook_data_validation', inputs=validation_inputs, targets=validation_targets)
np.savez('Audiobook_data_train', inputs=train_inputs, targets=train_targets)
```
